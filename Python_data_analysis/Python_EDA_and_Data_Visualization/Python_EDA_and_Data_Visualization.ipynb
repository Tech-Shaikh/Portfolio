```python
airbnb_data = pd.read_csv('datasets.csv',encoding_errors='ignore')
```

Task 4 : Data Cleaning

```python
airbnb_data.isnull().sum()
```

Based on the data type of the column handling the missing(for integer column updating as mean and for string droping out)

```python

column = airbnb_data.columns
for col in column:
    if airbnb_data[col].dtype in ('int','float'):
        airbnb_data[col] = airbnb_data[col].fillna(airbnb_data[col].mean())
    else:
        airbnb_data.dropna(subset=[col],inplace=True)
```

Task 4 : Data Analysis


**Univariant Analysis**

```python
#Dealing with the outlier 
airbnb_df = airbnb_data[airbnb_data['price']<1500]
sns.boxplot(data=airbnb_df,x='price')
```

```python
#price Distribution
plt.figure(figsize=(8,5))
plt.title("Price Distribution")
sns.histplot(data=airbnb_df,x='price',bins=100)
plt.ylabel("Frequency")
plt.show()
```

```python
#availability  Distribution
plt.figure(figsize=(6,3))
plt.title("Availability Distribution")
sns.histplot(data=airbnb_df,x='availability_365')
plt.ylabel("Frequency")
plt.show()
```

```python
airbnb_data.groupby(by='neighbourhood_group').agg({'price':'mean'})
```

Feature Engineering

```python
# adding new column price per bed

airbnb_data['price_per_bed'] = airbnb_data['price']/airbnb_data['beds']
```

```python
airbnb_data.columns
```

```python

# price dependency on neighbourhood
sns.barplot(data=airbnb_data, x='neighbourhood_group', y='price', hue='room_type')
```

```python
# number of reviews and price rel
plt.figure(figsize=(8, 5))
plt.title("Locality and Review Dependency")
sns.scatterplot(data=airbnb_df, x='number_of_reviews', y='price', hue='neighbourhood_group')
plt.show()
```

```python
sns.pairplot(data=airbnb_df, vars=['price', 'minimum_nights', 'number_of_reviews', 'availability_365'], hue='room_type')
```

```python
#Geographical Distribution of AirBnb Listing
plt.figure(figsize=(10, 7))
sns.scatterplot(data=airbnb_df, x='longitude', y='latitude', hue='room_type')
plt.title("Geographical Distribution of AirBnb Listing")
plt.show()
```

```python
corr = airbnb_df[['latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month', 'availability_365', 'beds']].corr()
corr

plt.figure(figsize=(8, 6))
sns.heatmap(data=corr, annot=True)
```
